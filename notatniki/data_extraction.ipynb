```python
import pandas as pd
```

```python
df = pd.read_csv('../dane/fra_cleaned_updated.csv', sep=';', encoding='utf-8')
df.head()
```

```python
accord_columns = ['mainaccord1', 'mainaccord2', 'mainaccord3', 'mainaccord4', 'mainaccord5']

unique_accords_set = set()

for col in accord_columns:
    unique_elements = df[col].dropna().unique()
    unique_accords_set.update(unique_elements)

unique_accords_list = sorted(list(unique_accords_set))

print(f"Znaleziono {len(unique_accords_list)} unikalnych akordów.")
```

```python
note_columns = ['Top', 'Middle', 'Base']

unique_notes_set = set()

for col in note_columns:
    
    valid_rows = df[col].dropna()
    
    for text_block in valid_rows:
        split_notes = text_block.split(',')
        cleaned_notes = [note.strip() for note in split_notes if note.strip() != '']
        unique_notes_set.update(cleaned_notes)

unique_notes_list = sorted(list(unique_notes_set))

print(f"Wyodrębniono {len(unique_notes_list)} unikalnych nut.")

for note in unique_notes_list[100:120]:
    print(f"- {note}")
```

```python
unique_notes_set = set()
for col in ['Top', 'Middle', 'Base']:
    valid_rows = df[col].dropna()
    for text_block in valid_rows:
        split_notes = text_block.split(',')
        cleaned_notes = [note.strip() for note in split_notes if note.strip() != '']
        unique_notes_set.update(cleaned_notes)

final_unique_notes = sorted(list(unique_notes_set))
print(f"Liczba unikalnych nut: {len(final_unique_notes)}")

unique_accords_set = set()
for col in ['mainaccord1', 'mainaccord2', 'mainaccord3', 'mainaccord4', 'mainaccord5']:
    unique_accords_set.update(df[col].dropna().unique())

final_unique_accords = sorted(list(unique_accords_set))
print(f"Liczba unikalnych akordów: {len(final_unique_accords)}")

df_notes = pd.DataFrame(final_unique_notes, columns=['Note Name'])
df_accords = pd.DataFrame(final_unique_accords, columns=['Accord Name'])

df_notes.to_csv('unique_specific_notes.csv', sep=';', index=False, encoding='utf-8')
df_accords.to_csv('unique_accords.csv', sep=';', index=False, encoding='utf-8')
```
